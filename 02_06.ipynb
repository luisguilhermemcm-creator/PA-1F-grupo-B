```python
#Fazer um jogo de adivinhação

#O usuário vai adivinhar um número de 0 até 10.

#Regras

#O usuário tera 03 chances para adivinhar
#Informar se o palpite do usuário é maior ou menor que o número.
#Imprimir mensagem de está quente morno ou frio.

import random

a = random.randint(0, 10)

numero = int(input('Digite um número de 0 a 10: '))
chances = 3

while True:
  if numero == a:
    print ('Você acertou')
    break

  if chances == 0:
    print ('Você perdeu')
    print (f'O número secreto era {a}.')
    break

  if numero > a:
    print (f"O numero {numero} é maior que o número secreto")
  elif numero < a:
    print (f"O numero {numero} é menor que o número secreto")

  chances -= 1
  print (f"Você ainda tem {chances} chances")
  numero = int(input('Digite um número de 0 a 10: '))
```
