```python
#1
#Par ou ímpar
#Peça um número ao usuário e informe se ele é par ou ímpar

# Entrada
numero = int(input("Digite um número: "))

#Processamento

if numero  %2 == 0:
  print (f"O número {numero} é par")

else:
  print (f"O número {numero} é ímpar")
```

```python
#2
#Solicite dois números e mostre qual é o maior (ou se são iguais).

#Entrada

numero1 = int(input("Digite o primeiro número: "))
numero2 = int(input("Digite o segundo número: "))

#Processamento

if numero1 > numero2 :
  print (f"O primeiro número é maior")

elif numero2 > numero1 :
  print (f"o segundo numero é maior")

else:
  print (f"Os números são iguais")
```

```python
#3
#Peça a nota de um aluno e informe:
#Aprovado (nota ≥ 7)
#Recuperação (5 ≤ nota < 7)
#Reprovado (nota < 5)

#Entrada
nota = float(input("Digite a nota do aluno: "))

#Processamento

if nota <5 :
  print (f"O aluno foi reprovado")

elif nota >= 5 and nota < 7 :
  print (f"O aluno está de recuperação")

else:
  print (f"O aluno foi aprovado")
```

```python
#4
#Receba a idade e classifique:
#Criança (0–12)
#Adolescente (13–17)
#Adulto (18–59)
#Idoso (60+)

#Entrada

idade = int(input("Digite a idade: "))

#Processamento

if idade >= 0 and idade <= 12:
  print (f"Criança")

elif idade >=13 and idade <= 17:
  print (f"Adolescente")

elif idade >= 18 and idade <= 59:
  print (f"Adulto")

else:
  print (f"Idoso")
```

```python
#6
# Peça um número e diga se ele é positivo, negativo ou zero.

#Entrada

numero = int(input("Digite um número: "))

#Processamento

if numero > 0:
  print (f"O número é positivo")

elif numero < 0:
  print (f"O número é negativo")

else:
  print (f"O número é zero")
```

```python
#7
#Peça dois números e uma operação (+, -, *, /). Use if/elif para calcular o resultado.

#Entrada
numero1 = int(input("Digite o primeiro número: "))
operação = input("Digite a operação (+, -, *, /): ")
numero2 = int(input("Digite o segundo número: "))
#Processamento

if operação == "+":
  resultado1 = numero1 + numero2
  print (f"O resultado da soma é: {resultado1}")

elif operação == "-":
  resultado2 = numero1 - numero2
  print (f"O resultado da subtração é: {resultado2}")

elif operação == "*":
  resultado3 = numero1 * numero2
  print (f"O resultado da multiplicação é: {resultado3}")

elif operação == "/":
  resultado4 = numero1 / numero2
  print (f"O resultado da divisão é: {resultado4}")

else:
  print (f"Operação inválida")
```

```python
#8
#Peça um número de 1 a 7 e mostre o dia correspondente (ex: 1 = domingo).

#entrada
dia = int(input("Digite um número de 1 a 7: "))

#Processamento

if dia == 1:
  print (f"O dia é domingo")

elif dia == 2:
  print (f"O dia é segunda")

elif dia == 3:
  print (f"O dia é terça")

elif dia == 4:
  print (f"O dia é quarta")

elif dia == 5:
  print (f"O dia é quinta")

elif dia == 6:
  print (f"O dia é sexta")

elif dia == 7:
  print (f"O dia é sábado")

else:
  print (f"O dia não existe")
```

```python
#10
#Peça o valor de uma compra:
#≥ 100: 10% de desconto
#≥ 50 e < 100: 5% de desconto
#< 50: sem desconto

#Entrada

valor = float(input("Digite o valor da compra: "))

#Processamento

if valor < 50 :
  print (f"O valor da compra é: {valor}  sem desconto")

elif valor > 50 and valor <100 :
  resultado1 = valor * 0.95
  print (f"O valor da compra é: {resultado1} com desconto de 5%")

elif valor >100 :
  resultado2 = valor * 0.9
  print (f"O valor da compra é: {resultado2} com desconto de 10%")
```
