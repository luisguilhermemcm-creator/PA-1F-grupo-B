for - geralmente é utilizado quando sabemos quantas vezes algo vai se repetir
while - é usado quando não sabemos

```python
senha = ""

while senha != 1234:
  senha = int(input("Digite a senha: "))
print("Senha inválida")

print ("Acesso liberado")
```

```python
#Escrever um programa que faz uma contagem regressiva de 10 até 0

for i in range (10, -1, -1):
  print (i)
```

```python
#Escrever um programa que recebe 04 notas referente a cada bimestre e calcule a média das notas
#e imprimi a média do aluno
#e se ele foi aprovado, recuperação ou reprovado.

#Utilizar while
#Definir uma variável para o bimestre e uma variavel para nota e uma variavel para o resultado

nota = 0
bimestre = 1
resultado = 0

while bimestre <= 4:
  nota=(float(input(f"Digite a nota do bimestre {bimestre}: ")))
  bimestre += 1

resultado = resultado + nota
#Processamento

print("A média do aluno é: ", resultado)

while resultado < 5:
  print("Reprovado")
  break
while resultado >= 5 and resultado < 7:
  print("Recuperação")
  break
while resultado >= 7:
  print("Aprovado")
  break
```
