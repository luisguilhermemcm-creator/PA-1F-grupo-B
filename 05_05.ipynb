```python
# 1 Escreva um programa que o usuário digita um número e o programa imprime até
#O número digitado

numero = int(input("Digite um número: "))

for i in range(1, numero + 1):
    print(i)
```

```python
# 3 Escreva um programa que o usuário digita um número e o programa imprime os números pares

numero = int(input("Digite um número: "))

for i in range(2, numero + 1, 2):
    print(i)
```

```python
# 4 Escreva um programa que o usuário digita um número e o programa imprime os números impares

numero = int(input("Digite um número "))

for i in range(1, numero + 1, 2):
  print(i)
```

```python
# 5 Escreva um programa que o usuário digita um número e o programa soma os numeros até o número digitado
numero = int(input("Digite um número: "))

soma = 0
for i in range(1, numero + 1):
    soma += i
print("A soma dos números de 1 a", numero, "é:", soma)
```

```python
# 6 Escreva um programa que o usuário digita uma palavra e o programa conta as letras das palavras

palavra = input ("Digite uma palavra ")

contador = 0
for letra in palavra:
    contador += 1
print("A palavra tem", contador, "letras")
```

```python
# 7 Escreva um programa que o usuário digita uma palavra e o programa conta as vogais das palavras

palavra = input("Digite uma palavra: ")

contador = 0
for letra in palavra:
    if letra in "aeiouAEIOU":
        contador += 1
print("A palavra tem", contador, "vogais")
```

```python
# 8 Escreva um programa que o usuário digita uma palavra e o programa imprime a palavra invertida

palavra = str (input("Digite uma palavra:"))

invertida = ""
for letra in palavra:
    invertida = letra + invertida
print("A palavra invertida é:", invertida)
```

```python
# 9 Escreva um programa que o usuário digita 04 números e o programa imprime o maior

numero1 = int(input("Digite número 1 : "))
numero2 = int(input("Digite número 2: "))
numero3 = int(input("Digite número 3: "))
numero4 = int(input("Digite número 4 :"))

maior = numero1
if numero2 > maior:
    maior = numero2
if numero3 > maior:
    maior = numero3
if numero4 > maior:
    maior = numero4
print("O maior número é:", maior)
```
